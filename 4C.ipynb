```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.layers import Dense , Embedding , Lambda
import random 
import re
```

```python
text = """Deep learning is subset of machine learning.
machine learning uses activation function for learning non linear problems."""
```

```python
corpus = text.lower().replace("." , "").split()
print(corpus)
print("Total Words : " , len(corpus))
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus)
word2idx = tokenizer.word_index
idx2word = tokenizer.index_word
vocab_size = len(word2idx) + 1
print(vocab_size)
```

```python
window_size = 3 
data = []

for i in range(window_size , len(corpus) - window_size):
    context = word2idx[corpus[i - 2]] , word2idx[corpus[i - 1]] , word2idx[corpus[i + 1]] , word2idx[corpus[i + 2]]
    target = word2idx[corpus[i]]
    data.append((context , target))

contexts = np.array([x[0] for x in data])
targets = np.array([x[1] for x in data])

print("Sample Training Data : " , contexts)
```

```python
model = keras.Sequential([
    Embedding(vocab_size , 200),
    Lambda(lambda x: tf.reduce_mean(x , axis=1)),
    Dense(vocab_size , activation = "softmax")
])
```

```python
model.compile(optimizer="adam" , loss="sparse_categorical_crossentropy" , metrics = ["accuracy"])
```

```python
model.fit(contexts , targets , epochs=30 , verbose = 1)
```
