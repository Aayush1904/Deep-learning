```python
# 2B. Implement the Image classification CNN model for classifying hand-written
# MNIST dataset by dividing the model into following 4 stages:
# a. Loading and preprocessing the image data
# b. Defining the model's architecture
# c. Training the model
# d. Estimating the model's performance
```

```python
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow import keras
```

```python
train_df = pd.read_csv("/Users/aayushagarwal/Downloads/LP IV Updated Datasets/FNN and CNN/HandWritten MNIST/mnist_train.csv")
test_df = pd.read_csv("/Users/aayushagarwal/Downloads/LP IV Updated Datasets/FNN and CNN/HandWritten MNIST/mnist_test.csv")
```

```python
x_train = train_df.drop("label" , axis=1).values.reshape(-1 , 28 , 28 , 1)
y_train = train_df["label"]
x_test = test_df.drop("label" , axis=1).values.reshape(-1 , 28 , 28 , 1)
y_test = test_df["label"]
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
y_train = keras.utils.to_categorical(y_train , 25)
y_test = keras.utils.to_categorical(y_test , 25)
```

```python
model = keras.Sequential([
    keras.layers.Input(shape = (28 , 28 ,1)),
    keras.layers.Conv2D(32 , (3,3) , activation = "relu"),
    keras.layers.MaxPooling2D(pool_size = (2,2)),
    keras.layers.Flatten(),
    keras.layers.Dense(64 , activation = "relu"),
    keras.layers.Dense(25 , activation = "softmax")
])
```

```python
model.compile(optimizer="adam" , loss="categorical_crossentropy" , metrics=["accuracy"])
```

```python
model.fit(x_train , y_train , epochs = 11 , batch_size=32 , validation_split = 0.2)
```

```python
loss , accuracy = model.evaluate(x_test , y_test , verbose = 0)
print(loss)
print(accuracy)
```
